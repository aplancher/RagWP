```python
import ollama
```

```python
sentences=[
    "What year did Lebron James get drafted?",
    "Who are some of the best point guards this year?",
    "Who's NBA contract is ending in 2027?",
    "Who makes the most money in the NBA right now?",
    "What age do players usually retire from the NBA?",
    "What is a forward deployed ai engineer?",
    "Can an undergraduate who just graduated get a job as a forward deployed ai engineer?",
    "What skills do you need to become a forward deployed ai engineer?",
    "What is linear algebra?",
    "How do you find the determinant of a matrix?"

]

resp = ollama.embed(model='nomic-embed-text', input=sentences)
```

```python
print(len(resp['embeddings']))
embedding=resp['embeddings']
```

```python
import numpy as np

# store the sentence embeddings into NumPy matrix
matrix = np.array(embedding)
```

```python
query = "How much do NBA stars earn?"
q_vector = ollama.embed(model='nomic-embed-text', input= query)
q_vector= q_vector['embeddings']
q_vector=np.array(q_vector)
```

```python
np.linalg.norm(matrix, axis=1)
```

```python
np.linalg.norm(q_vector)
```

```python
q_vector = q_vector.T
q_vector.shape
```

```python
matrix.shape
```

```python
similarity= matrix @ q_vector
```
